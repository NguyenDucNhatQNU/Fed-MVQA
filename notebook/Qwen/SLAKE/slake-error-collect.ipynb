```python
import os
import gc
import time
import torch
import re
import numpy as np
import pandas as pd
from PIL import Image
import urllib.request
import zipfile
from datasets import load_dataset
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor
from peft import LoraConfig, get_peft_model
from torch.utils.data import DataLoader
from sklearn.metrics import accuracy_score, f1_score
import evaluate
from torch.optim.lr_scheduler import CosineAnnealingLR
from tqdm.auto import tqdm
import warnings

warnings.filterwarnings("ignore", category=UserWarning)

# ==========================================
# 0. THIẾT LẬP MÔI TRƯỜNG KAGGLE & CUDNN
# ==========================================
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "0"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
torch.backends.cudnn.benchmark = True 

runName = "FedMVQA_SLAKE_Qwen2VL2B_LoRA_448_Alpha10_R1_E2_Batch8"

kaggleBaseDir = "/kaggle/working/Fed-MVQA-Project"
dataDir = os.path.join(kaggleBaseDir, "data", "slake")
logDir = os.path.join(kaggleBaseDir, "logs")
checkpointDir = os.path.join(kaggleBaseDir, "checkpoints", runName)

os.makedirs(dataDir, exist_ok=True)
os.makedirs(logDir, exist_ok=True)
os.makedirs(checkpointDir, exist_ok=True)

logPath = os.path.join(logDir, f"{runName}_log.csv")
checkpointPath = os.path.join(checkpointDir, "best_fed_mvqa_model.pt")

predictionsPath = os.path.join(logDir, f"{runName}_predictions.csv")
qualitativePath = os.path.join(logDir, f"{runName}_qualitative_samples.txt")

# ==========================================
# 1. CÔNG CỤ QUẢN LÝ BỘ NHỚ VÀ ĐO LƯỜNG
# ==========================================
def clearMemory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

def getPeakReservedMemory():
    if torch.cuda.is_available():
        reservedGb = torch.cuda.memory_reserved() / (1024 ** 3)
        return f"{reservedGb:.2f} GB"
    return "0.00 GB"

def calculateTrainableParameters(model):
    trainableParams = 0
    allParams = 0
    for _, param in model.named_parameters():
        allParams += param.numel()
        if param.requires_grad:
            trainableParams += param.numel()
    percentage = 100 * trainableParams / allParams
    return trainableParams, allParams, percentage

def calculateCommunicationMetrics(loraWeights, totalParams, numClients):
    peftBytes = sum(tensor.numel() * tensor.element_size() for key, tensor in loraWeights.items())
    peftRoundCostMb = (peftBytes * 2 * numClients) / (1024 * 1024) 
    fullModelBytes = totalParams * 2 
    fullRoundCostMb = (fullModelBytes * 2 * numClients) / (1024 * 1024)
    percentageReduction = ((fullRoundCostMb - peftRoundCostMb) / fullRoundCostMb) * 100
    return peftRoundCostMb, fullRoundCostMb, percentageReduction

# ==========================================
# 2. CHUẨN BỊ DỮ LIỆU ĐẶC THÙ SLAKE
# ==========================================
print("\n--- CHUẨN BỊ DỮ LIỆU SLAKE (NON-IID α=1.0, 3 CLIENTS) ---")

if not os.path.exists("./imgs"):
    print(" [Hệ thống] Đang tải file ảnh imgs.zip từ Hugging Face...")
    urllib.request.urlretrieve("https://huggingface.co/datasets/BoKelvin/SLAKE/resolve/main/imgs.zip", "imgs.zip")
    with zipfile.ZipFile("imgs.zip", 'r') as zipRef:
        zipRef.extractall("./")
    os.remove("imgs.zip")
    print(" [Hệ thống] Đã chuẩn bị xong thư mục ảnh cục bộ!")

dataset = load_dataset("BoKelvin/SLAKE", cache_dir=dataDir)
englishTrain = dataset["train"].filter(lambda x: x["q_lang"] == "en")
englishTest = dataset["test"].filter(lambda x: x["q_lang"] == "en")

shuffledTrain = englishTrain.shuffle(seed=42)
testDataset = englishTest.shuffle(seed=42).select(range(min(500, len(englishTest))))

def partitionDataNonIidDirichlet(datasetInput, numClients=3, alpha=1.0):
    labels = [str(item["answer"]).strip().lower() for item in datasetInput]
    uniqueLabels = list(set(labels))
    clientIndices = {i: [] for i in range(numClients)}
    for label in uniqueLabels:
        labelIdxs = [i for i, l in enumerate(labels) if l == label]
        np.random.shuffle(labelIdxs)
        proportions = np.random.dirichlet([alpha] * numClients)
        splits = np.split(labelIdxs, (np.cumsum(proportions)[:-1] * len(labelIdxs)).astype(int))
        for i in range(numClients):
            clientIndices[i].extend(splits[i].tolist())
            
    clientDatasets = []
    for i in range(numClients):
        np.random.shuffle(clientIndices[i])
        clientDatasets.append(datasetInput.select(clientIndices[i]))
    return clientDatasets

clientDatasets = partitionDataNonIidDirichlet(shuffledTrain, numClients=3, alpha=1.0)
datasetClientA = clientDatasets[0]
datasetClientB = clientDatasets[1]
datasetClientC = clientDatasets[2]
dataSizes = [len(datasetClientA), len(datasetClientB), len(datasetClientC)]

print(f"Dữ liệu Client A: {len(datasetClientA)} mẫu")
print(f"Dữ liệu Client B: {len(datasetClientB)} mẫu")
print(f"Dữ liệu Client C: {len(datasetClientC)} mẫu")
print(f"Dữ liệu Global Test: {len(testDataset)} mẫu")

def loadSlakeImage(imgName):
    imgPath = os.path.join(".", "imgs", imgName.replace("imgs/", ""))
    return Image.open(imgPath).convert("RGB")

# ==========================================
# 3. KHỞI TẠO MÔ HÌNH TOÀN CỤC (LORA 448x448)
# ==========================================
print("\n--- KHỞI TẠO HẠ TẦNG FED-MVQA (LORA 7 MODS - 448x448 - FP16) ---")
clearMemory()

processor = AutoProcessor.from_pretrained(
    "Qwen/Qwen2-VL-2B-Instruct",
    min_pixels=256 * 28 * 28,  
    max_pixels=256 * 28 * 28 
)

baseModel = Qwen2VLForConditionalGeneration.from_pretrained(
    "Qwen/Qwen2-VL-2B-Instruct",
    torch_dtype=torch.float16, 
    device_map={"": 0}, 
    attn_implementation="sdpa"
)

baseModel.config.use_cache = False
baseModel.gradient_checkpointing_enable()

loraConfig = LoraConfig(
    r=32,
    lora_alpha=64,
    lora_dropout=0.1, 
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    task_type="CAUSAL_LM",
    bias="none"
)

globalPeftModel = get_peft_model(baseModel, loraConfig)
trainableParams, totalModelParams, trainPct = calculateTrainableParameters(globalPeftModel)
print(f"[Metrics] Trainable Params (LoRA): {trainableParams:,} (~{trainPct:.3f}%)")

def extractLoraWeights(model):
    return {k: v.cpu().clone() for k, v in model.state_dict().items() if "lora" in k}

globalWeights = extractLoraWeights(globalPeftModel)

# ==========================================
# 4. THIẾT LẬP DATA LOADER
# ==========================================
headerTokens = processor.tokenizer.encode("<|im_start|>assistant\n", add_special_tokens=False)
headerLen = len(headerTokens)

systemInstruction = (
    "You are an expert medical diagnostic assistant. Your task is to answer clinical questions "
    "based EXCLUSIVELY on the provided medical image.\n"
    "Strictly follow these rules:\n"
    "1. Analyze the provided [Target Case Image] and the [Question].\n"
    "2. Provide a concise and definitive answer (1-3 words if possible)."
)

def customCollateFn(batch):
    texts, images = [], []
    for item in batch:
        img = loadSlakeImage(item["img_name"])
        q = item["question"]
        a = str(item["answer"]).strip()
        
        messages = [
            {"role": "system", "content": [{"type": "text", "text": systemInstruction}]},
            {"role": "user", "content": [
                {"type": "text", "text": "[Target Case Image]\n"},
                {"type": "image", "image": img}, 
                {"type": "text", "text": f"\nCurrent Question: {q}\n\nBased on the visual evidence, what is the correct answer for the current case?\nFinal Answer:"}
            ]},
            {"role": "assistant", "content": [{"type": "text", "text": a}]}
        ]
        textPrompt = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=False)
        texts.append(textPrompt)
        images.append(img)

    inputs = processor(text=texts, images=images, return_tensors="pt", padding=True)
    labels = inputs["input_ids"].clone()
    for i in range(labels.shape[0]):
        tokenIds = labels[i].tolist()
        if processor.tokenizer.pad_token_id is not None:
            labels[i][labels[i] == processor.tokenizer.pad_token_id] = -100
        matchIdx = -1
        for j in range(len(tokenIds) - headerLen + 1):
            if tokenIds[j:j+headerLen] == headerTokens:
                matchIdx = j + headerLen 
                break
        if matchIdx != -1:
            labels[i, :matchIdx] = -100 
    inputs["labels"] = labels
    return inputs

batchSize = 8
trainLoaderA = DataLoader(datasetClientA, batch_size=batchSize, shuffle=True, collate_fn=customCollateFn, num_workers=2, pin_memory=True)
trainLoaderB = DataLoader(datasetClientB, batch_size=batchSize, shuffle=True, collate_fn=customCollateFn, num_workers=2, pin_memory=True)
trainLoaderC = DataLoader(datasetClientC, batch_size=batchSize, shuffle=True, collate_fn=customCollateFn, num_workers=2, pin_memory=True)

# ==========================================
# 5. HÀM HUẤN LUYỆN CỤC BỘ 
# ==========================================
def trainLocalClient(model, dataLoader, clientId, roundIdx, epochs=2, learningRate=5e-5): 
    model.train()
    decayFactor = 0.9 ** (roundIdx - 1)
    activeLr = learningRate * decayFactor
    optimizer = torch.optim.AdamW(model.parameters(), lr=activeLr, weight_decay=0.01)
    accumulationSteps = 1 
    scheduler = CosineAnnealingLR(optimizer, T_max=epochs * len(dataLoader) // accumulationSteps)
    
    print(f" [Client {clientId}] Bắt đầu LoRA 448x448 (Batch=8, AdamW, LR={activeLr:.2e})...")
    startTime = time.time()
    totalLoss = 0
    
    for epoch in range(epochs):
        optimizer.zero_grad(set_to_none=True)
        progressLoop = tqdm(dataLoader, desc=f"Client {clientId} | Epoch {epoch + 1}/{epochs}", leave=True, dynamic_ncols=True)
        
        for step, batch in enumerate(progressLoop):
            batch = {k: v.to(model.device) for k, v in batch.items()}
            with torch.amp.autocast("cuda", dtype=torch.float16):
                outputs = model(**batch)
                loss = outputs.loss / accumulationSteps
            loss.backward()
            
            if (step + 1) % accumulationSteps == 0 or (step + 1) == len(dataLoader):
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()
                scheduler.step()
                optimizer.zero_grad(set_to_none=True)
                
            stepLoss = loss.item() * accumulationSteps
            totalLoss += stepLoss
            currentLr = scheduler.get_last_lr()[0]
            progressLoop.set_postfix(Loss=f"{stepLoss:.4f}", LR=f"{currentLr:.2e}")
            
            del outputs, loss, batch
            
        avgLoss = totalLoss / len(dataLoader)
        print(f"      Epoch {epoch + 1} hoàn tất | Avg Loss: {avgLoss:.4f}")
    
    trainDuration = time.time() - startTime
    finalLoss = totalLoss / (len(dataLoader) * epochs)
    peakVram = getPeakReservedMemory()
    del optimizer, scheduler
    clearMemory()
    return extractLoraWeights(model), trainDuration, finalLoss, peakVram

# ==========================================
# 6. THUẬT TOÁN FEDERATED AVERAGING
# ==========================================
def simpleFederatedAveraging(globalWeightsInput, clientWeightsList, dataSizes):
    totalSamples = sum(dataSizes)
    newGlobalWeights = {} 
    keys = list(globalWeightsInput.keys())
    for key in keys:
        if "lora" in key:
            weightedSum = torch.zeros_like(globalWeightsInput[key], dtype=torch.float32, device="cpu")
            for i in range(len(clientWeightsList)):
                weightRatio = dataSizes[i] / totalSamples
                weightedSum += clientWeightsList[i][key].float() * weightRatio
            newGlobalWeights[key] = weightedSum.to(globalWeightsInput[key].dtype)
            
        for i in range(len(clientWeightsList)):
            if key in clientWeightsList[i]:
                del clientWeightsList[i][key]
    gc.collect()
    return newGlobalWeights

# ==========================================
# 7. HÀM ĐÁNH GIÁ VÀ XUẤT CSV PREDICTION
# ==========================================
def cleanOpenAnswer(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    stopwords = {'a', 'an', 'the', 'is', 'are', 'was', 'were', 'it', 'there', 'shows', 'showing', 'of', 'in', 'on', 'to'}
    words = [w for w in words if w not in stopwords]
    return " ".join(words)

def classifyErrorCategory(question, gt, pred):
    q = question.lower()
    if gt == "no" and pred == "yes":
        return "Hallucination"
    if any(w in q for w in ["organ", "where", "location", "part", "lobe", "side", "left", "right", "kidney", "liver", "heart"]):
        return "Anatomical understanding"
    if any(w in q for w in ["disease", "abnormal", "finding", "diagnos", "condition", "tumor", "nodule", "fracture"]):
        return "Disease identification"
    if any(w in q for w in ["reason", "why", "suggest", "cause", "evidence"]):
        return "Medical reasoning"
    if any(w in q for w in ["modality", "type", "mri", "x-ray", "xray", "ct", "ultrasound", "bigger", "smaller", "contain"]) or (gt == "yes" and pred == "no"):
        return "Visual recognition"
    return "Question interpretation"

def evaluateGlobalModel(model, datasetInput, processorInput, exportCsvPath=None):
    model.eval()
    print(f"\n [Server] Đang đánh giá trên tập Test ({len(datasetInput)} mẫu)...")
    
    openPreds, openRefs, closePreds, closeRefs = [], [], [], []
    overallCorrect, inferenceTimes = 0, []
    predictionRecords = []
    
    yesKeywords = ['yes', 'correct', 'true', 'positive', 'present', 'visible']
    noKeywords = ['no', 'incorrect', 'false', 'negative', 'absent', 'normal', 'none']
    
    evalLoop = tqdm(datasetInput, desc="Evaluating", leave=False, dynamic_ncols=True)
    
    for item in evalLoop:
        img = loadSlakeImage(item["img_name"])
        q = item["question"]
        gtAnswer = str(item["answer"]).strip().lower()
        ansType = str(item.get("answer_type", "")).strip().upper()
        if ansType == "": ansType = "CLOSED" if gtAnswer in ["yes", "no"] else "OPEN"
        
        messages = [
            {"role": "system", "content": [{"type": "text", "text": systemInstruction}]},
            {"role": "user", "content": [
                {"type": "text", "text": "[Target Case Image]\n"},
                {"type": "image", "image": img}, 
                {"type": "text", "text": f"\nCurrent Question: {q}\n\nBased on the visual evidence, what is the correct answer for the current case?\nFinal Answer:"}
            ]}
        ]
        
        textPrompt = processorInput.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        inputs = processorInput(text=[textPrompt], images=[img], return_tensors="pt", padding=True)
        inputs = {k: v.to(model.device) for k, v in inputs.items()}
        
        startTime = time.time()
        with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.float16):
            generatedIds = model.generate(**inputs, max_new_tokens=15, do_sample=False)
        inferenceTimes.append(time.time() - startTime)
        
        generatedIdsTrimmed = [outIds[len(inIds):] for inIds, outIds in zip(inputs["input_ids"], generatedIds)]
        aiAnswerRaw = processorInput.batch_decode(generatedIdsTrimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False)[0].strip().lower()
        aiAnswer = aiAnswerRaw.replace("final answer:", "").strip()
        
        isCorrectLabel = "Incorrect"
        aiAnswerClean = cleanOpenAnswer(aiAnswer)
        gtAnswerClean = cleanOpenAnswer(gtAnswer)
        
        if ansType == "CLOSED":
            if gtAnswer in ['yes', 'no']:
                hasYes = any(word in aiAnswer for word in yesKeywords)
                hasNo = any(word in aiAnswer for word in noKeywords)
                if hasYes and not hasNo: predClean = 'yes'
                elif hasNo and not hasYes: predClean = 'no'
                else:
                    firstYesPos = aiAnswer.find('yes') if 'yes' in aiAnswer else 9999
                    firstNoPos = aiAnswer.find('no') if 'no' in aiAnswer else 9999
                    predClean = 'yes' if firstYesPos < firstNoPos else ('no' if firstNoPos < firstYesPos else 'unknown')
            else:
                predClean = aiAnswerClean
                
            closePreds.append(predClean)
            closeRefs.append(gtAnswerClean if gtAnswer not in ['yes', 'no'] else gtAnswer)
            
            if predClean == (gtAnswerClean if gtAnswer not in ['yes', 'no'] else gtAnswer): 
                overallCorrect += 1
                isCorrectLabel = "Correct"
        else:
            if aiAnswerClean == gtAnswerClean: 
                overallCorrect += 1
                isCorrectLabel = "Correct"
            elif gtAnswerClean != "" and (gtAnswerClean in aiAnswerClean or aiAnswerClean in gtAnswerClean):
                isCorrectLabel = "Partially Correct"
            openPreds.append(aiAnswerClean); openRefs.append(gtAnswerClean)
                
        errorCategory = "None"
        if isCorrectLabel != "Correct":
            errorCategory = classifyErrorCategory(q, gtAnswerClean, aiAnswerClean)

        imgPathFormat = os.path.join(".", "imgs", item["img_name"].replace("imgs/", ""))
        predictionRecords.append({
            "Image_Path": imgPathFormat,
            "Question": q,
            "Ground_Truth": gtAnswer,
            "Fed_MVQA_Prediction": aiAnswer,
            "Question_Type": ansType,
            "Status": isCorrectLabel,
            "Error_Category": errorCategory
        })
            
        del inputs, generatedIds, generatedIdsTrimmed
        torch.cuda.empty_cache()

    if exportCsvPath is not None:
        pd.DataFrame(predictionRecords).to_csv(exportCsvPath, index=False, encoding='utf-8-sig')
        print(f" [Hệ thống] Đã xuất toàn bộ kết quả dự đoán ra: {exportCsvPath}")

    res = {"overallAccuracy": overallCorrect / len(datasetInput) if len(datasetInput) > 0 else 0, "avgInferenceTime": sum(inferenceTimes) / len(inferenceTimes)}
    if closeRefs: res["Closed"] = {"ACC": accuracy_score(closeRefs, closePreds), "F1": f1_score(closeRefs, closePreds, average='weighted', zero_division=0), "count": len(closeRefs)}
    if openRefs:
        bertRes = evaluate.load("bertscore").compute(predictions=openPreds, references=openRefs, lang="en")
        res["Open"] = {"BLEU-1": evaluate.load("bleu").compute(predictions=openPreds, references=[[r] for r in openRefs], max_order=1)["bleu"], "ROUGE-L": evaluate.load("rouge").compute(predictions=openPreds, references=[[r] for r in openRefs])["rougeL"], "BERTScore": sum(bertRes["f1"]) / len(bertRes["f1"]), "count": len(openRefs)}
    
    gc.collect()
    return res

# ==========================================
# 8. VÒNG LẶP HỌC LIÊN KẾT CHÍNH 
# ==========================================
maxRounds = 1  
localEpochs = 2
cumulativePeftGb = 0.0
performanceLog = []
peakVramRecorded = "0.00 GB"

print(f"\n=== BẮT ĐẦU HUẤN LUYỆN FED-MVQA SLAKE ({maxRounds} ROUNDS, {localEpochs} EPOCHS, ALPHA=1.0, BATCH=8) ===")

for roundIdx in range(1, maxRounds + 1):
    print(f"\n" + "="*50)
    print(f"--- COMMUNICATION ROUND {roundIdx} ---")
    print("="*50)
    latestTrainTimes = []
    
    globalPeftModel.load_state_dict(globalWeights, strict=False)
    localWeightsA, timeA, lossA, vramA = trainLocalClient(globalPeftModel, trainLoaderA, "A", roundIdx, localEpochs, 5e-5)
    latestTrainTimes.append(timeA)
    
    globalPeftModel.load_state_dict(globalWeights, strict=False)
    localWeightsB, timeB, lossB, vramB = trainLocalClient(globalPeftModel, trainLoaderB, "B", roundIdx, localEpochs, 5e-5)
    latestTrainTimes.append(timeB)
    
    globalPeftModel.load_state_dict(globalWeights, strict=False)
    localWeightsC, timeC, lossC, vramC = trainLocalClient(globalPeftModel, trainLoaderC, "C", roundIdx, localEpochs, 5e-5)
    latestTrainTimes.append(timeC)
    peakVramRecorded = vramC
    
    print("\n [Server] Đang tổng hợp trọng số LoRA (FedAvg IN-RAM)...")
    globalWeights = simpleFederatedAveraging(globalWeights, [localWeightsA, localWeightsB, localWeightsC], dataSizes)
    
    globalPeftModel.load_state_dict(globalWeights, strict=False)
    
    currentCsvPath = predictionsPath if roundIdx == maxRounds else None
    evalResults = evaluateGlobalModel(globalPeftModel, testDataset, processor, exportCsvPath=currentCsvPath)
    
    roundAccuracy = evalResults.get("overallAccuracy", 0) * 100
    roundAccClosed = evalResults.get("Closed", {}).get("ACC", 0) * 100
    roundF1Closed = evalResults.get("Closed", {}).get("F1", 0) * 100
    roundBleu = evalResults.get("Open", {}).get("BLEU-1", 0)
    roundRouge = evalResults.get("Open", {}).get("ROUGE-L", 0)
    roundBert = evalResults.get("Open", {}).get("BERTScore", 0)
    
    peftRoundMb, fullRoundMb, reductionPct = calculateCommunicationMetrics(globalWeights, totalModelParams, 3)
    cumulativePeftGb += (peftRoundMb / 1024)
    avgTrainingTime = sum(latestTrainTimes) / len(latestTrainTimes) if len(latestTrainTimes) > 0 else 0
    averageRoundLoss = (lossA + lossB + lossC) / 3
    
    roundLog = {
        "Round": roundIdx,
        "Avg Loss": round(averageRoundLoss, 4),
        "Overall Acc (%)": round(roundAccuracy, 2),
        "Closed Acc (%)": round(roundAccClosed, 2),
        "Closed F1 (%)": round(roundF1Closed, 2),
        "Open BLEU-1": round(roundBleu, 4),
        "Open ROUGE-L": round(roundRouge, 4),
        "Open BERTScore": round(roundBert, 4),
        "MB per Round": round(peftRoundMb, 2),
        "Avg Train Time (s)": round(avgTrainingTime, 2),
        "Peak VRAM": peakVramRecorded,
        "Cumul Data (GB)": round(cumulativePeftGb, 4)
    }
    performanceLog.append(roundLog)
    
    print(f"\n [KẾT QUẢ VÒNG {roundIdx}]")
    print(f"  + Loss Trung Bình  : {averageRoundLoss:.4f}")
    print(f"  + Overall Accuracy : {roundAccuracy:.2f}%")
    if "Closed" in evalResults:
        print(f"  + Closed Questions : Acc = {roundAccClosed:.2f}%, F1 = {roundF1Closed:.2f}%")
    if "Open" in evalResults:
        print(f"  + Open Questions   : BLEU-1 = {roundBleu:.4f}, ROUGE-L = {roundRouge:.4f}, BERTScore = {roundBert:.4f}")
    print(f"  + Hiệu Suất        : Train = {avgTrainingTime:.2f}s | VRAM = {peakVramRecorded} | Truyền tải = {peftRoundMb:.2f} MB")

    pd.DataFrame(performanceLog).to_csv(logPath, index=False)
    
    if roundIdx == maxRounds:
        torch.save(globalWeights, checkpointPath)

# ==========================================
# 9. XUẤT BÁO CÁO TỔNG KẾT
# ==========================================
print("\n" + "="*80)
print(f" FINAL SUMMARY REPORT: FED-MVQA (SLAKE, LORA 448x448, BATCH 8, FP16) ")
print("="*80)
print(f"  + Trainable Params       : {trainableParams:,} (~{trainPct:.3f}%)")
print(f"  + Overall Test Accuracy  : {evalResults['overallAccuracy']*100:.2f}%")
if "Closed" in evalResults:
    print(f"  + Closed Questions ({evalResults['Closed']['count']} mẫu) : Acc = {roundAccClosed:.2f}%, F1 = {roundF1Closed:.2f}%")
if "Open" in evalResults:
    print(f"  + Open Questions ({evalResults['Open']['count']} mẫu)   : BLEU-1 = {roundBleu:.4f}, ROUGE-L = {roundRouge:.4f}, BERTScore = {roundBert:.4f}")
print("="*80)

# ==========================================
# 10. TRÍCH XUẤT VÍ DỤ PHÂN TÍCH ĐỊNH TÍNH (QUALITATIVE ANALYSIS)
# ==========================================
print("\n" + "="*80)
print(" [QUALITATIVE ANALYSIS] TRÍCH XUẤT MẪU DỰ ĐOÁN ĐỂ PHÂN TÍCH LỖI ")
print("="*80)

try:
    df_preds = pd.read_csv(predictionsPath)
    
    correct_cases = df_preds[df_preds['Status'] == 'Correct']
    incorrect_cases = df_preds[df_preds['Status'].isin(['Incorrect', 'Partially Correct'])]
    
    sampled_correct = correct_cases.sample(1, random_state=42) if len(correct_cases) > 0 else pd.DataFrame()
    
    unique_errors = incorrect_cases['Error_Category'].unique()
    sampled_incorrect_list = []
    for error_type in unique_errors:
        if len(sampled_incorrect_list) >= 3: break
        sampled_incorrect_list.append(incorrect_cases[incorrect_cases['Error_Category'] == error_type].sample(1))
    
    sampled_incorrect = pd.concat(sampled_incorrect_list) if sampled_incorrect_list else incorrect_cases.sample(3, random_state=42)
    
    qualitative_text = "=== QUALITATIVE ANALYSIS SAMPLES ===\n\n"
    
    if not sampled_correct.empty:
        row = sampled_correct.iloc[0]
        qualitative_text += "--------------------------------------------------\n"
        qualitative_text += "[CASE 1] - CORRECT PREDICTION\n"
        qualitative_text += f"Image Path : {row['Image_Path']}\n"
        qualitative_text += f"Question   : {row['Question']}\n"
        qualitative_text += f"Ground Truth: {row['Ground_Truth']}\n"
        qualitative_text += f"Fed-MVQA Pred: {row['Fed_MVQA_Prediction']}\n"
        qualitative_text += "--------------------------------------------------\n\n"
        
    for idx, row in enumerate(sampled_incorrect.itertuples(), start=2):
        qualitative_text += "--------------------------------------------------\n"
        qualitative_text += f"[CASE {idx}] - {row.Status.upper()}\n"
        qualitative_text += f"Actual Error Category : >> {row.Error_Category} <<\n"
        qualitative_text += f"Image Path : {row.Image_Path}\n"
        qualitative_text += f"Question   : {row.Question}\n"
        qualitative_text += f"Ground Truth: {row.Ground_Truth}\n"
        qualitative_text += f"Fed-MVQA Pred: {row.Fed_MVQA_Prediction}\n"
        qualitative_text += "--------------------------------------------------\n\n"
        
    print(qualitative_text)
    
    with open(qualitativePath, "w", encoding="utf-8") as f:
        f.write(qualitative_text)
    print(f" -> Đã lưu các ví dụ định tính ra file Text để dán vào Word: {qualitativePath}")
    
except Exception as e:
    print(f" Không thể trích xuất ví dụ. Lỗi: {str(e)}")
```
